## 🔹 Paso 2: Analizar si el negocio es rentable

### 2.1 Cálculo de KPIs principales

**🎯 Objetivo:** Calcular los indicadores clave del negocio para evaluar ingresos, costos y rentabilidad.

Se usan los 3 datasets (`orders`, `catalog`, `marketing_spend`):

**📊 Parte 1: Rentabilidad del negocio**
- ¿Cuál es el ingreso total (revenue)? 
- ¿Cuál es el costo total? 
- ¿Cuánto se ha invertido en marketing? 
- ¿El negocio es rentable? (calcular profit)  

---

**📈 Parte 2: Comportamiento de ventas**
- ¿Cuál es el ticket promedio por orden? 
- ¿Cuál es la cantidad promedio de productos por orden? 
- ¿Cuál es el producto más vendido?
- ¿Cuánto se ha gastado en marketing por canal? 

```python
#KPIs Rentabilidad del negocio 
#Calculo de ingreso total 
revenue_total = orders["monto_total"].sum()
print(f"Revenue total: ${revenue_total:,.2f}")

# Incorporación el costo unitario del catálogo
ventas = orders.merge(catalog[["nombre_producto", "costo_unitario"]],on="nombre_producto",how="left")

#Validación del merge 
print("Filas originales:", orders.shape[0])
print("Filas después del merge:", ventas.shape[0])
print("Costos ausentes:", ventas["costo_unitario"].isna().sum())

# Calculo el costo total
costo_total = ( ventas["cantidad"] * ventas["costo_unitario"]).sum()
print(f"Costo total: ${costo_total:,.2f}")

#Calculo inversión total en marketing
gasto_marketing = marketing["gasto"].sum()
print(f"Inversión total en marketing: "f"${gasto_marketing:,.2f}")

#Calculo Profit(rentabilidad)
profit = revenue_total - costo_total - gasto_marketing
print(f"Profit: ${profit:,.2f}")
```

- Durante el periodo analizado, el negocio generó un revenue total de 9,610,018.94. El costo de los productos vendidos fue de 3,828,869.01 y la inversión en marketing ascendió a 2,871,843.53. Después de descontar ambos conceptos, se obtuvo un profit positivo de 2,909,306.40. Por lo tanto, el negocio fue rentable durante el periodo analizado.

```python
#KPIs Comportamiento de ventas
#Calculo ticket promedio 
ticket_promedio = (revenue_total/ orders["id_pedido"].nunique())
print(f"Ticket promedio: ${ticket_promedio:,.2f}")

#Calculo cantidad promedio de producto por orden 
productos_promedio = (orders.groupby("id_pedido")["cantidad"].sum().mean())
print("Cantidad promedio de productos por orden: "f"{productos_promedio:.2f}")


#Calculo producto más vendido 
# Unidades vendidas por producto
ventas_por_producto = (orders.groupby("nombre_producto")["cantidad"].sum().sort_values(ascending=False))
# Producto más vendido
producto_mas_vendido = ventas_por_producto.idxmax()
cantidad_producto_mas_vendido = ventas_por_producto.max()
print(f"Producto más vendido: {producto_mas_vendido}")
print("Unidades del producto más vendido: "f"{cantidad_producto_mas_vendido:,.0f}")

#Calculo gasto de marketing por canal 
marketing_por_canal = (marketing.groupby("canal")["gasto"].sum().sort_values(ascending=False))
print("\nGasto de marketing por canal:")
display(marketing_por_canal.to_frame(name="gasto_total"))
```

- El ticket promedio por pedido fue de 385.85.
- Cada pedido incluyó, en promedio, 1.50 unidades de productos.
- El producto más vendido fue Vacuum-Pro-Black, con 6,284 unidades vendidas.
- El canal con mayor inversión fue social, con 976,818.37, seguido de organic con 972,650.96 y paid_search con 922,374.20.

```python
# Explorar tabla events
# =========================
query_events = '''
SELECT *
FROM events;
'''
events = pd.read_sql(query_events, con=engine)
events.head()
```

```python
# PARTE 1: Totales del funnel
# ======================

query_totals = '''
SELECT
    nombre_evento,
    COUNT(DISTINCT id_usuario) AS usuarios_unicos
FROM events
GROUP BY nombre_evento
ORDER BY
    CASE nombre_evento
        WHEN 'first_visit' THEN 1
        WHEN 'select_item' THEN 2
        WHEN 'add_to_cart' THEN 3
        WHEN 'begin_checkout' THEN 4
        WHEN 'add_payment_info' THEN 5
        WHEN 'purchase' THEN 6
        ELSE 7
    END;
'''

totals = pd.read_sql(query_totals, con=engine)
totals
```

```python
# PARTE 2: Conversiones
# ======================

query_conversion = '''
WITH funnel AS (
    SELECT
        nombre_evento,
        COUNT(DISTINCT id_usuario) AS usuarios_unicos,

        CASE nombre_evento
            WHEN 'first_visit' THEN 1
            WHEN 'select_item' THEN 2
            WHEN 'add_to_cart' THEN 3
            WHEN 'begin_checkout' THEN 4
            WHEN 'add_payment_info' THEN 5
            WHEN 'purchase' THEN 6
            ELSE 7
        END AS orden_etapa

    FROM events
    GROUP BY nombre_evento
),

funnel_ordenado AS (
    SELECT
        nombre_evento,
        usuarios_unicos,
        orden_etapa,

        LAG(usuarios_unicos) OVER (
            ORDER BY orden_etapa
        ) AS usuarios_etapa_anterior,

        FIRST_VALUE(usuarios_unicos) OVER (
            ORDER BY orden_etapa
        ) AS usuarios_inicio

    FROM funnel
)

SELECT
    nombre_evento AS etapa,
    usuarios_unicos,

    usuarios_etapa_anterior - usuarios_unicos
        AS usuarios_perdidos,

    ROUND(
        usuarios_unicos::decimal
        / NULLIF(usuarios_etapa_anterior, 0) * 100,
        2
    ) AS tasa_conversion_paso

FROM funnel_ordenado
ORDER BY orden_etapa;
'''

conversion = pd.read_sql(query_conversion, con=engine)
conversion
```

Análisis del funnel de conversión

- El funnel comenzó con 7,796 usuarios en first_visit y finalizó con 6,240 usuarios en purchase. Esto representa una tasa de conversión final del 80.04%.

- La mayor pérdida se produjo entre begin_checkout y add_payment_info: la cantidad de usuarios disminuyó de 7,208 a 6,250, lo que representa una pérdida de 958 usuarios. La conversión entre estas etapas fue del 86.71%, por lo que el proceso de incorporación de la información de pago constituye el principal punto de abandono.

- Entre select_item y add_to_cart se observaron 52 usuarios adicionales, produciendo una conversión del 100.69%. Esto indica que el conteo agrupa usuarios únicos por evento de manera independiente y no garantiza que cada usuario haya completado todos los pasos anteriores. Algunos usuarios pudieron acceder directamente al carrito o no tener registrado el evento select_item.

- En la última transición, de add_payment_info a purchase, la conversión fue del 99.84%, lo que indica que casi todos los usuarios que ingresaron sus datos de pago terminaron la compra.

```python
# Retención por cohortes
# ======================

query_cohort_retention_final = '''
WITH actividad_usuarios AS (
    SELECT
        u.id_usuario,
        DATE_TRUNC(
            'month',
            CAST(u.fecha_registro AS DATE)
        )::date AS cohorte,
        CAST(ua.fecha_actividad AS DATE) AS fecha_actividad,
        ua.dias_despues_registro,
        ua.activo
    FROM users AS u
    LEFT JOIN user_activity AS ua
        ON u.id_usuario = ua.id_usuario
),

retencion AS (
    SELECT
        cohorte,
        COUNT(DISTINCT id_usuario) AS clientes_iniciales,

        COUNT(DISTINCT CASE
            WHEN dias_despues_registro = 7
                 AND activo = 1
            THEN id_usuario
        END) AS retenido_w1,

        COUNT(DISTINCT CASE
            WHEN dias_despues_registro = 14
                 AND activo = 1
            THEN id_usuario
        END) AS retenido_w2,

        COUNT(DISTINCT CASE
            WHEN dias_despues_registro = 21
                 AND activo = 1
            THEN id_usuario
        END) AS retenido_w3

    FROM actividad_usuarios
    GROUP BY cohorte
)

SELECT
    cohorte,
    clientes_iniciales,
    retenido_w1,
    retenido_w2,
    retenido_w3,

    ROUND(
        retenido_w1::decimal /
        NULLIF(clientes_iniciales, 0) * 100,
        2
    ) AS semana_1,

    ROUND(
        retenido_w2::decimal /
        NULLIF(clientes_iniciales, 0) * 100,
        2
    ) AS semana_2,

    ROUND(
        retenido_w3::decimal /
        NULLIF(clientes_iniciales, 0) * 100,
        2
    ) AS semana_3

FROM retencion
ORDER BY cohorte;
'''

# Ejecutar la consulta
cohorte_final = pd.read_sql(query_cohort_retention_final, con=engine)
cohorte_final
```

La retención semanal se mantiene relativamente estable entre las cohortes analizadas, con valores cercanos al 42% durante las primeras tres semanas posteriores al registro.

- La retención promedio fue aproximadamente de 42.01% en la semana 1, 41.96% en la semana 2 y 41.92% en la semana 3.
- La cohorte de enero de 2025 presentó la retención más alta en la semana 1, con 42.84%.
- La cohorte de abril de 2025 alcanzó la mayor retención en la semana 2, con 43.40%.
- La cohorte de febrero de 2025 registró la mayor retención en la semana 3, con 43.98%.

## 🔹 Paso 5: Validar si los cambios generan impacto (test estadístico)

🎯 **Objetivo:** Evaluar si la modificación en la UI del checkout impacta la **tasa de conversión de compra**.

---

1. **Analizar el dataset** `experiment_checkout_ui.csv` para identificar la métrica principal **conversion**.
   - La métrica **conversion** es 1 si el usuario completó la compra, 0 si no.    
2. **Plantear la hipótesis estadística**     
3. **Aplicar el test estadístico adecuado** 
4. **Interpretar el resultado**  

---
Hipótesis estadística
   - **H₀ (Hipótesis nula):** No existe diferencia en la tasa de conversión de compra entre el grupo con la interfaz original y el grupo con la interfaz modificada. 
   - **H₁ (Hipótesis alternativa):** Existe diferencia en la tasa de conversión de compra entre el grupo con la interfaz original y el grupo con la interfaz modificada. 
   
**Test estadístico:** Prueba Z  
**Nivel de significancia alpha:** 0.05

```python
# Cargar y analizar el dataset experimento
experiment = pd.read_csv('https://practicum-content.s3.amazonaws.com/datasets/experiment_checkout_ui.csv')
display(experiment.head())
```

```python
from statsmodels.stats.proportion import proportions_ztest

# Contar número de conversiones y total de usuarios
conversiones = experiment.groupby("variante")["convirtio"].sum()
totales = experiment.groupby("variante")["convirtio"].count()

# Pasar los valores a formato lista
exitos = [
    conversiones["control"],
    conversiones["tratamiento"]
]

observaciones = [
    totales["control"],
    totales["tratamiento"]
]

# Mostrar las tasas de conversión
tasa_control = conversiones["control"] / totales["control"]
tasa_tratamiento = (
    conversiones["tratamiento"] /
    totales["tratamiento"]
)

print(f"Tasa de conversión del control: {tasa_control:.2%}")
print(
    "Tasa de conversión del tratamiento: "
    f"{tasa_tratamiento:.2%}"
)

# Aplicar prueba Z de dos proporciones
z_stat, p_value = proportions_ztest(
    exitos,
    observaciones
)

print(f"\nEstadístico z: {z_stat:.4f}")
print(f"Valor p: {p_value:.4f}")

# Interpretar los resultados
alpha = 0.05

if p_value < alpha:
    print(
        "Rechazamos la hipótesis nula: existe evidencia "
        "estadísticamente significativa de una diferencia "
        "en la tasa de conversión."
    )
else:
    print(
        "No rechazamos la hipótesis nula: no existe evidencia "
        "estadística suficiente de una diferencia en la tasa "
        "de conversión."
    )
```

```python
# Obtener tasa o porcentaje de conversión
tasa_control = exitos[0] / observaciones[0]
tasa_tratamiento = exitos[1] / observaciones[1]

print(
    f"Tasa de conversión del grupo control: "
    f"{tasa_control:.2%}"
)

print(
    f"Tasa de conversión del grupo tratamiento: "
    f"{tasa_tratamiento:.2%}"
)

# Interpretar la dirección de los resultados
if tasa_control > tasa_tratamiento:
    diferencia = tasa_control - tasa_tratamiento

    print(
        f"\nEl grupo control tiene una mayor tasa de conversión. "
        f"La diferencia es de {diferencia:.2%}."
    )

elif tasa_tratamiento > tasa_control:
    diferencia = tasa_tratamiento - tasa_control

    print(
        f"\nEl grupo tratamiento tiene una mayor tasa de conversión. "
        f"La diferencia es de {diferencia:.2%}."
    )

else:
    print(
        "\nAmbos grupos tienen la misma tasa de conversión."
    )
```

- El grupo control obtuvo una tasa de conversión del 15.69%, mientras que el grupo tratamiento alcanzó una tasa del 16.29%. Descriptivamente, la interfaz modificada presentó una conversión 0.60 puntos porcentuales mayor.

- Sin embargo, la prueba Z de dos proporciones produjo un estadístico de −0.8133 y un valor p de 0.4161. Debido a que el valor p es mayor que el nivel de significancia de 0.05, no se rechaza la hipótesis nula.

- Por lo tanto, no existe evidencia estadística suficiente para afirmar que la modificación de la interfaz del checkout haya generado un cambio real en la tasa de conversión. Aunque el tratamiento mostró una conversión ligeramente superior, la diferencia observada no es estadísticamente significativa.
